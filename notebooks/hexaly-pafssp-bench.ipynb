# PAFSSP - Hexaly Benchmark

```python
from bnbprob.pafssp import CallbackBnB, PermFlowShop
from bnbprob.pafssp.hxly import (
    solve_hexaly,
    solve_hexaly_fs,
)
from bnbprob.pafssp.instances import random_parallel_semilines
from bnbprob.pafssp.mip.grb_positional import GurobiWrapper
```

## Define a random instance

```python
instance_data = random_parallel_semilines(10, (5, 5), 12)
```

```python
instance_data.edges
```

## Flow-shop formulation

From template: https://www.hexaly.com/templates/flow-shop-problem

```python
hx_fs_sol = solve_hexaly_fs(instance_data, 60)
```

## Job-shop-like formulation

Template from: https://www.hexaly.com/templates/job-shop-scheduling-problem-jssp

```python
hx_js_sol = solve_hexaly(instance_data, 60)
```

## Analytical solution (B&B)

```python
problem = PermFlowShop.from_p(instance_data.p, instance_data.edges)
bnb = CallbackBnB(problem)
bnb_sol = bnb.solve()
print(bnb_sol)
```

## MILP Model

Position-based formulation

```python
mip_model = GurobiWrapper(instance_data.p, instance_data.edges)
mip_model.set_param("TimeLimit", 60)
mip_sol = mip_model.solve()
print(mip_sol)
```
